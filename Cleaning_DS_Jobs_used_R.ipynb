**Replace Duplicated Values in Job Title**

```r
# Senior Data Scientist
df$Job.Title = gsub(
  "^(Sr\\.?|Senior)\\.? Data Scientist.*",
  "Senior Data Scientist",
  df$Job.Title,
  ignore.case = TRUE
)
# Data Scientist / Machine Learning
df$Job.Title = gsub(
  "^Data Scientist[ /-]+Machine Learning.*",
  "Data Scientist - Machine Learning",
  df$Job.Title,
  ignore.case = TRUE
)
# Data Engineer
df$Job.Title = gsub(
  "^Jr\\.? Data Engineer.*",
  "Junior Data Engineer",
  df$Job.Title,
  ignore.case = TRUE
)
df$Job.Title = gsub(
  "^Sr\\.? Data Engineer.*",
  "Senior Data Engineer",
  df$Job.Title,
  ignore.case = TRUE
)
df$Job.Title = gsub(
  "^(Senior|Sr\\.?) Data Analyst.*$",
  "Senior Data Analyst",
  df$Job.Title,
  ignore.case = TRUE
)
df$Job.Title = gsub(
  "^Data Modeler.*$",
  "Data Modeler",
  df$Job.Title
)
df$Job.Title = gsub(
  "^\\(Sr\\.\\) Data Scientist.*$",
  "Senior Data Scientist",
  df$Job.Title
)
df$Job.Title = gsub(
  "^Staff Data Scientist.*$",
  "Staff Data Scientist",
  df$Job.Title
)
df$Job.Title = gsub(
  "^Principal Data Scientist.*$",
  "Principal Data Scientist",
  df$Job.Title
)
df$Job.Title = gsub(
  "^Lead Data Scientist.*$",
  "Lead Data Scientist",
  df$Job.Title
)
df$Job.Title = gsub(
  "^Data Engineer.*$",
  "Data Engineer",
  df$Job.Title
)
df$Job.Title = gsub(
  "^Business Intelligence Analyst I-.*$",
  "Business Intelligence Analyst",
  df$Job.Title
)
df$Job.Title = gsub(
  "^Machine Learning Engineer, Sr\\.$",
  "Senior Machine Learning Engineer",
  df$Job.Title
)
df$Job.Title = gsub(
  "^Jr\\. Business Data Analyst.*$",
  "Junior Business Data Analyst",
  df$Job.Title
)
df$Job.Title = gsub(
  "^Data Scientist \\(TS/SCI.*\\)$|^Data Scientist - TS/SCI.*$",
  "Data Scientist",
  df$Job.Title
)
df$Job.Title = gsub(
  "^Analytics Manager.*$",
  "Analytics Manager",
  df$Job.Title
)
df$Job.Title = gsub(
  "^Say Business Data Analyst$",
  "Business Data Analyst",
  df$Job.Title
)
df$Job.Title = gsub(
  "^Machine Learning Scientist / Engineer$|^Machine Learning Engineer/Scientist$",
  "Machine Learning Scientist / Engineer",
  df$Job.Title
)
df$Job.Title = gsub(
  "^Health Plan Data Analyst, Sr$",
  "Senior Data Analyst",
  df$Job.Title
)
df$Job.Title = gsub(
  "^Data Scientist - Contract$",
  "Data Scientist",
  df$Job.Title
)

unique(df$Job.Title)
```

**Replace Values in Salary column**

```r
df$Salary.Estimate=gsub(" \\(Glassdoor est\\.\\)","",df$Salary.Estimate)
df$Salary.Estimate=gsub(" \\(\\)", "", df$Salary.Estimate)
df$Salary.Estimate=gsub("\\(Employer est\\.\\)","",df$Salary.Estimate)
unique(df$Salary.Estimate)
```

```r
df$Salary.Midpoint <- sapply(
  strsplit(gsub("\\$|K", "", df$Salary.Estimate), "-"),
  function(x) mean(as.numeric(x)) * 1000
)
#head(df[, c("Salary.Estimate", "Salary.Midpoint")], 10)
unique(df$Salary.Midpoint)
```

**Exploratory Data Analysis Questions (Answer in R)**


```r
#answar1
#1. What are the top 5 most common Job Titles for data science positions?
df %>%
  count(Job.Title,sort=TRUE) %>%
  head(5)
```

```r
#answar2
#2. Which city or location has the highest number of job postings?
df %>%
  count(Location,sort=TRUE) %>%
  head(1)
```

```r
#answar3
#3. What is the average company rating across all job postings?
  mean(df$Rating,na.rm = TRUE)
```

```r
#answar4
#4. Which industries have the highest average salaries?
df %>%
  group_by(Industry) %>%
  summarise(average_Salary=mean(Salary.Midpoint,na.rm = TRUE)) %>%
  arrange(desc(average_Salary)) %>%
  head(5)
```

```r
#answar5
#5. Is there a relationship between company rating and salary estimate?
# Answer 5
# Is there a relationship between company rating and salary estimate?
#كدة مفيش علاقه corrilation=0
cor(df$Rating, df$Salary.Midpoint, use = "complete.obs")
```

```r
#answar6
#6. What percentage of job postings mention Python or R in the job description?
mentions_python_r = grepl("Python|\\bR\\b", df$Job.Description, ignore.case = TRUE)
percentage = mean(mentions_python_r, na.rm = TRUE) * 100
percentage
```

```r
#answar7
#7. Which type of ownership (Private, Public, Government) posts the most jobs?
df %>%
  count(Type.of.ownership,sort = TRUE) #%>%
  #head(5)
```

```r
#answar8
#8. How many companies were founded before 2000 and are still hiring data roles?
df %>%
  filter(Founded<2000) %>%
  count(Company.Name,sort=TRUE) %>%
  head(5)
```

```r
#answar9
#9. What is the most common industry sector among data-related jobs?
df %>%
  count(Sector,sort=TRUE) %>%
  head(1)
```

**Insights**

1. **Data Scientist is the most common job title** in the dataset, with 246 job postings. This suggests that Data Scientists are highly represented among the available data-related positions.

2. **Information Technology is the most common industry sector**, with 176 job postings. This suggests that data scientists may find many opportunities in the IT sector.

3. **Company rating has almost no linear relationship with salary estimate**, with a correlation of approximately 0.014. This suggests that a higher company rating does not necessarily mean a higher salary in this dataset.
